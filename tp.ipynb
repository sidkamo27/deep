```python
import pandas as pd
import numpy as np
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding , Lambda , Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import to_categorical
```

```python
with open("cbow_text.txt" , "r") as file:
    text = file.read()

tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

data = tokenizer.texts_to_sequences([text])[0]
vocab = len(tokenizer.word_index) + 1

print("Total Words: " , len(data))
print("Vocab Size" , vocab)
```

```python
window_size = 2
context = []
target = []

for i in range(window_size , len(data) - window_size):
    context.append(data[i-2:i] + data[i+1 : i+3])
    target.append(data[i])

context = np.array(context)
target = to_categorical(target , vocab)


print("Context Shape: " , context.shape)
print("Target Shape: " , target.shape)
```

```python
model = Sequential([
    Embedding(vocab , 100 , input_length = 4),
    Lambda(lambda x : tf.reduce_mean(x , axis = 1)),
    Dense(vocab , activation="softmax")
])
```

```python
model.compile(
    optimizer="adam",
    loss = "categorical_crossentropy",
    metrics = ["accuracy"]
)
```

```python
model.fit(context , target , epochs=100 , verbose=1)
```
